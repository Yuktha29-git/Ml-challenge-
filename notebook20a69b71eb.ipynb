```python
# Create sets of source IDs for fast lookup
src2_id_set = set(df_src2["entity_id"])
src3_id_set = set(df_src3["entity_id"])

missing_s2 = 0
missing_s3 = 0
checked_s2 = 0
checked_s3 = 0

chunk_size = 100_000

for start in range(0, len(df_truth), chunk_size):
    chunk = df_truth["matched_entity_ids"].iloc[start:start + chunk_size].dropna()

    for value in chunk:
        ids = value.split(",")

        for entity_id in ids:
            if entity_id.startswith("S2-"):
                checked_s2 += 1
                if entity_id not in src2_id_set:
                    missing_s2 += 1

            elif entity_id.startswith("S3-"):
                checked_s3 += 1
                if entity_id not in src3_id_set:
                    missing_s3 += 1

print("Checked Source 2 matches:", checked_s2)
print("Missing Source 2 IDs:", missing_s2)

print("\nChecked Source 3 matches:", checked_s3)
print("Missing Source 3 IDs:", missing_s3)
```

```python
# Take a small sample of Source 1
sample = df_src1.sample(n=10000, random_state=42)

# Build a lookup from Source 2: (business_name, country) -> number of records
src2_name_country = (
    df_src2.groupby(["business_name", "country"], dropna=False)
           .size()
)

# Check how many Source 1 records have an exact name + country match in Source 2
keys = list(zip(sample["business_name"], sample["country"]))

exact_matches = [
    (name, country) in src2_name_country.index
    for name, country in keys
]

print("Sample Source 1 rows:", len(sample))
print("Exact name + country matches in Source 2:", sum(exact_matches))
print("Percentage:", round(sum(exact_matches) / len(sample) * 100, 2), "%")
```

```python
# Build a lookup using exact name + address + country
src2_name_address_country = (
    df_src2.groupby(
        ["business_name", "business_address", "country"],
        dropna=False
    ).size()
)

keys = list(zip(
    sample["business_name"],
    sample["business_address"],
    sample["country"]
))

exact_matches_address = [
    key in src2_name_address_country.index
    for key in keys
]

print("Sample Source 1 rows:", len(sample))
print(
    "Exact name + address + country matches in Source 2:",
    sum(exact_matches_address)
)
print(
    "Percentage:",
    round(sum(exact_matches_address) / len(sample) * 100, 2),
    "%"
)
```

```python
# Show a few Source 1 records that have an exact name + country match
matched_sample = sample[exact_matches].head(5)

for _, row in matched_sample.iterrows():
    print("\nSOURCE 1")
    print("Name   :", row["business_name"])
    print("Address:", row["business_address"])
    print("Country:", row["country"])

    candidates = df_src2[
        (df_src2["business_name"] == row["business_name"]) &
        (df_src2["country"] == row["country"])
    ][["entity_id", "business_name", "business_address", "country"]].head(3)

    print("\nSOURCE 2 CANDIDATES")
    print(candidates.to_string(index=False))
```

```python
# Count how many Source 2 records share each (business_name, country)
src2_name_counts = (
    df_src2.groupby(["business_name", "country"], dropna=False)
           .size()
)

# Look at the 10,000 Source 1 sample
sample_candidate_counts = []

for _, row in sample.iterrows():
    key = (row["business_name"], row["country"])
    sample_candidate_counts.append(
        src2_name_counts.get(key, 0)
    )

sample_candidate_counts = pd.Series(sample_candidate_counts)

print("No Source 2 candidate:",
      int((sample_candidate_counts == 0).sum()))

print("Exactly 1 candidate:",
      int((sample_candidate_counts == 1).sum()))

print("2-5 candidates:",
      int(sample_candidate_counts.between(2, 5).sum()))

print("More than 5 candidates:",
      int((sample_candidate_counts > 5).sum()))

print("\nMaximum candidates for one Source 1 record:",
      int(sample_candidate_counts.max()))
```

```python
import re

def normalize_text(text):
    if pd.isna(text):
        return ""
    text = str(text).lower()
    text = re.sub(r"[^a-z0-9]+", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

sample["name_norm"] = sample["business_name"].map(normalize_text)
sample["address_norm"] = sample["business_address"].map(normalize_text)

print(sample[[
    "business_name",
    "name_norm",
    "business_address",
    "address_norm"
]].head(5).to_string(index=False))
```

```python
# Build normalized name + country lookup for Source 2
src2_name_norm_counts = (
    df_src2.assign(
        name_norm=df_src2["business_name"].map(normalize_text)
    )
    .groupby(["name_norm", "country"], dropna=False)
    .size()
)

# Count candidates for our 10,000 Source 1 sample
normalized_candidate_counts = []

for _, row in sample.iterrows():
    key = (row["name_norm"], row["country"])
    normalized_candidate_counts.append(
        src2_name_norm_counts.get(key, 0)
    )

normalized_candidate_counts = pd.Series(normalized_candidate_counts)

print("No normalized-name candidate:",
      int((normalized_candidate_counts == 0).sum()))

print("At least 1 normalized-name candidate:",
      int((normalized_candidate_counts > 0).sum()))

print("Exactly 1 candidate:",
      int((normalized_candidate_counts == 1).sum()))

print("2-5 candidates:",
      int(normalized_candidate_counts.between(2, 5).sum()))

print("More than 5 candidates:",
      int((normalized_candidate_counts > 5).sum()))

print("\nMaximum candidates:",
      int(normalized_candidate_counts.max()))
```

```python
from difflib import SequenceMatcher

row = sample.iloc[0]

print("SOURCE 1")
print("Name   :", row["business_name"])
print("Address:", row["business_address"])
print("Country:", row["country"])

candidates = df_src2[
    df_src2["country"] == row["country"]
].head(1000)

results = []

for _, candidate in candidates.iterrows():
    score = SequenceMatcher(
        None,
        row["name_norm"],
        normalize_text(candidate["business_name"])
    ).ratio() * 100

    results.append((
        score,
        candidate["entity_id"],
        candidate["business_name"],
        candidate["business_address"]
    ))

results.sort(reverse=True)

print("\nTop 10 name similarities:")

for score, entity_id, name, address in results[:10]:
    print(f"\nScore: {score:.1f}")
    print("ID:", entity_id)
    print("Name:", name)
    print("Address:", address)
```

```python
# Calculate address similarity for the same candidates

results_address = []

for _, candidate in candidates.iterrows():
    score = SequenceMatcher(
        None,
        row["address_norm"],
        normalize_text(candidate["business_address"])
    ).ratio() * 100

    results_address.append((
        score,
        candidate["entity_id"],
        candidate["business_name"],
        candidate["business_address"]
    ))

results_address.sort(reverse=True)

print("SOURCE 1 ADDRESS:")
print(row["business_address"])

print("\nTop 10 address similarities:")

for score, entity_id, name, address in results_address[:10]:
    print(f"\nScore: {score:.1f}")
    print("ID:", entity_id)
    print("Name:", name)
    print("Address:", address)
```

```python
# Get the ground-truth matches for our sample row

source1_id = row["entity_id"]

truth_row = df_truth[
    df_truth["source1_entity_id"] == source1_id
]["matched_entity_ids"].iloc[0]

print("Source 1 ID:", source1_id)
print("Source 1 name:", row["business_name"])
print("Source 1 address:", row["business_address"])

print("\nGround-truth matched IDs:")
print(truth_row)
```

```python
matched_ids = truth_row.split(",")

print("SOURCE 1")
print("ID      :", row["entity_id"])
print("Name    :", row["business_name"])
print("Address :", row["business_address"])
print("Country :", row["country"])

print("\nGROUND-TRUTH MATCHES")

for entity_id in matched_ids:
    if entity_id.startswith("S2-"):
        match = df_src2[df_src2["entity_id"] == entity_id]
    else:
        match = df_src3[df_src3["entity_id"] == entity_id]

    print("\n", match.to_string(index=False))
```

```python
def tokens(text):
    return set(normalize_text(text).split())

source1_address_tokens = tokens(row["business_address"])

print("SOURCE 1 ADDRESS TOKENS:")
print(source1_address_tokens)

print("\nGROUND-TRUTH ADDRESS OVERLAP:")

for entity_id in matched_ids:
    if entity_id.startswith("S2-"):
        match = df_src2[df_src2["entity_id"] == entity_id].iloc[0]
    else:
        match = df_src3[df_src3["entity_id"] == entity_id].iloc[0]

    match_tokens = tokens(match["business_address"])
    overlap = source1_address_tokens & match_tokens

    print(
        entity_id,
        "| shared tokens:", overlap,
        "| count:", len(overlap)
    )
```
